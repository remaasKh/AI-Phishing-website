# Phishing Website Detection using Random Forest

## Step 1: Import Libraries

```python
import pandas as pd
```

## Step 2: Extract the Dataset

```python
import zipfile

with zipfile.ZipFile("archive.zip", "r") as zip_ref:
    zip_ref.extractall()
```

## Step 3: Load the Dataset

```python
data = pd.read_csv("dataset.csv")
data.head()
```

## Step 4: Explore the Dataset

```python
data.info()
```

## Step 5: Define Features and Target

>



```python
X = data.drop("Result", axis=1)
y = data["Result"]
```

## Step 6: Split the Dataset

```python
from sklearn.model_selection import train_test_split

X = X.drop("index", axis=1)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.4,
    random_state=42
)
```

```python
X.shape
```

## Step 7: Build the Random Forest Model

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()
```

## Step 8: Train the Model

```python
model.fit(X_train, y_train)
```

## Step 9: Make Predictions

```python
y_pred = model.predict(X_test)
```

## Step 10: Evaluate the Model

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy * 100, 2), "%")
```

## Step 11: Generate the Classification Report

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

## Step 12: Confusion Matrix

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

## Step 13: Evaluate Training and Testing Performance

```python
y_train_pred = model.predict(X_train)

train_accuracy = accuracy_score(y_train, y_train_pred)

print(f"Training Accuracy: {train_accuracy * 100:.2f}%")
print(f"Testing Accuracy: {accuracy * 100:.2f}%")
```

## Step 14: Conclusion
The Random Forest model achieved a testing accuracy of 96.70%, indicating strong performance in detecting phishing websites. The training and testing accuracies are close (99.07% vs. 96.70%), suggesting that the model generalizes well and does not show significant overfitting or underfitting.
